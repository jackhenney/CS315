```python
#@title Lex program to count identifiers
import re

# identifier:
# starts with letter or _
# then only letters or _
pattern = r'^[a-zA-Z_][a-zA-Z_]*$'

count = 0

with open("program.txt", "r") as f:
    for line in f:
        word = line.strip()
        if re.match(pattern, word):
            print(word, "is valid")
            count += 1
        else:
            print(word, "is invalid")

print("Total valid identifiers:", count)
```
